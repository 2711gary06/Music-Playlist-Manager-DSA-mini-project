### Music Playlist Manager (C++)

```python
%%writefile playlist_manager.cpp
#include <iostream>
#include <string>
#include <vector>
#include <algorithm>

struct Song {
    std::string title;
    std::string artist;
    int duration;
};

class PlaylistManager {
private:
    std::vector<Song> playlist;
    int currentSongIndex = 0;

public:
    void addSong(const std::string& title, const std::string& artist, int duration) {
        playlist.push_back({title, artist, duration});
        std::cout << "Added: \"" << title << "\" by " << artist << "\n";
    }

    void displayPlaylist() const {
        if (playlist.empty()) {
            std::cout << "Playlist is empty.\n";
            return;
        }
        std::cout << "\n--- Current Playlist ---\n";
        for (size_t i = 0; i < playlist.size(); ++i) {
            std::cout << (i + 1) << ". " << playlist[i].title << " - " << playlist[i].artist
                      << " (" << playlist[i].duration / 60 << ":"
                      << (playlist[i].duration % 60 < 10 ? "0" : "") << playlist[i].duration % 60 << ")\n";
        }
    }

    void playNext() {
        if (playlist.empty()) {
            std::cout << "No songs to play.\n";
            return;
        }
        std::cout << "Playing: " << playlist[currentSongIndex].title << " by " << playlist[currentSongIndex].artist << "\n";
        currentSongIndex = (currentSongIndex + 1) % playlist.size();
    }
};

int main() {
    PlaylistManager manager;


    manager.addSong("Bohemian Rhapsody", "Queen", 354);
    manager.addSong("Blinding Lights", "The Weeknd", 200);
    manager.addSong("Hotel California", "Eagles", 390);

    manager.displayPlaylist();

    std::cout << "\n--- Simulating Player ---\n";
    manager.playNext();
    manager.playNext();

    return 0;
}
```

```python
!g++ -O3 playlist_manager.cpp -o playlist_manager
!./playlist_manager
```
